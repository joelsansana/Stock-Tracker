# Stock Analysis Example

This notebook demonstrates how to use the python_stocks modules.

```python
# Import modules
import sys
sys.path.insert(0, 'src')

from ark_fetcher import ARKDataFetcher
from stock_data import StockDataFetcher, TechnicalIndicators
import pandas as pd
import matplotlib.pyplot as plt
```

## ARK ETF Holdings

```python
# Fetch ARK ETF holdings
fetcher = ARKDataFetcher()
arkk = fetcher.get_holding('ARKK')
print(f"ARKK Holdings: {len(arkk)} stocks")
display(arkk.head())
```

## Stock Price Data

```python
# Get Apple stock data
stock_fetcher = StockDataFetcher()
aapl = stock_fetcher.get_price('AAPL', period='6mo')
print(f"AAPL Data: {len(aapl)} days")
display(aapl.tail())
```

## Technical Indicators

```python
# Calculate indicators
close = aapl['Close']

aapl['SMA_20'] = TechnicalIndicators.sma(close, 20)
aapl['SMA_50'] = TechnicalIndicators.sma(close, 50)
aapl['RSI'] = TechnicalIndicators.rsi(close)

macd, signal, hist = TechnicalIndicators.macd(close)
aapl['MACD'] = macd
aapl['MACD_Signal'] = signal

display(aapl.tail())
```

## Visualization

```python
# Plot price and SMA
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

axes[0].plot(aapl.index, aapl['Close'], label='Price')
axes[0].plot(aapl.index, aapl['SMA_20'], label='SMA 20')
axes[0].plot(aapl.index, aapl['SMA_50'], label='SMA 50')
axes[0].set_title('AAPL Price with SMA')
axes[0].legend()
axes[0].grid(True)

axes[1].plot(aapl.index, aapl['RSI'])
axes[1].axhline(y=70, color='r', linestyle='--')
axes[1].axhline(y=30, color='g', linestyle='--')
axes[1].set_title('RSI (14)')
axes[1].set_ylim(0, 100)
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
# Clean up
fetcher.close()
stock_fetcher.close()
```
